**Программа выполняет МОРФОСИНТАКСИЧЕСКИЙ РАЗБОР ТЕКСТА и записывает его в файл.
Для старта программы запустить ячейку.**

```
#ввести название файла с текстом для обработки
file = str(input('Введите название текстового файла (н-р, Рамаяна_3.txt): '))

#ввести путь к папке с файлами на гугл-диске
path = str(input('Укажите путь к папке с файлами (н-р, /content/drive/My Drive/Colab Notebooks/диплом/): '))

#подключение к гугл-диску
from google.colab import drive
drive.mount('/content/drive')

import re
import nltk
nltk.download('punkt')

#установка deeppavlov. Локально у меня не получилось установить, в колабе все устанавливается (4 строчки ниже).

!pip install git+https://github.com/andersjo/dependency_decoding
!pip install deeppavlov
!python -m deeppavlov install morpho_ru_syntagrus_pymorphy
!python -m deeppavlov install ru_syntagrus_joint_parsing

from deeppavlov import build_model
joint_model = build_model("ru_syntagrus_joint_parsing", download=True)


with open(path + file, 'r', encoding = 'utf-8') as r:
  text = r.read()

reg_space = re.compile('\s{2,}', re.DOTALL)
reg_punct = ("\s[0-9!\(\),-:;\?\[\]«»'–“„\.•…]+|[0-9!\(\),-:;\?\[\]«»'–“„\.•…]+\s|[0-9!\(\),-:;\?\[\]«»'–“„\.•…]+$|[\(\)«»'“„]+")
text = re.sub('\n', ' ', text)
text = re.sub('\s{2,}', ' ', text)
sents_number = len(nltk.sent_tokenize(text))

for indx, sent in enumerate(nltk.sent_tokenize(text)):
  try:
    print(indx)
    print('Обработано ' + str(indx+1) + ' предложений из ' + str(sents_number))
    if re.search('[а-яА-Я](-\s-[0-9]+-\s)[а-яА-Я]+', sent):
      br = re.search('[а-яА-Я](-\s-[0-9]+-\s)[а-яА-Я]+', sent).group(1)
      sent = re.sub(br, '', sent)

    text = re.sub(r'\d', '', sent)
    text = re.sub(r'\n', ' ', text)
    text = re.sub(r'[A-Za-z]', '', text)
    text = re.sub(reg_punct, ' ', text)

    if re.search(reg_space, text):
      text = re.sub(reg_space, ' ', text)

    predictions = joint_model([text])
    with open(path + 'deeppavlov_{}'.format(file), 'a', encoding = 'utf-8') as j:
      j.write('(')
      j.write(text)
      j.write(',')
      j.write(str(predictions))
      j.write(')')
      j.write('\n')
  except:
    continue
```
